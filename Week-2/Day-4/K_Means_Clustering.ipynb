Elbow Method

```python
rate=[]

k=range(2,10)

for i in k:
    kmeans=KMeans(n_clusters=i)
    kmeans.fit(df)
    rate.append(kmeans.inertia_)#kmeans.inertia_ gives the sum of squared distances of samples to their closest cluster center.
```

```python
rate
```

```python
plt.plot(k,rate,"red")
plt.xlabel("K")
plt.title("Elbow Method")
plt.show()
```
