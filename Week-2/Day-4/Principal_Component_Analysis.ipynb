Importing the Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import warnings
import matplotlib.pyplot as plt
warnings.filterwarnings("ignore")
```

Exporatory Data Analysis

```python
from sklearn.datasets import load_digits
digits=load_digits()
```

```python
digits["data"]
```

```python
digits["target"]
```

```python
digits["feature_names"]
```

```python
df_data=pd.DataFrame(digits["data"],columns=digits["feature_names"]) 
```

```python
df_data.head()
```

```python
df_data.shape
```

Preparing for PCA

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler= StandardScaler()
scaled_data=scaler.fit_transform(df_data)#Needs to be converted ino a dataframe structure
```

```python
df_new=pd.DataFrame(scaled_data,columns=digits["feature_names"])
```

```python
df_new.head()
```

PCA with two Components

```python
from sklearn.decomposition import PCA
```

```python
pca= PCA(n_components=2)#we need two components so n_components=2
```

```python
df_pca=pca.fit_transform(df_new)
```

```python
df_pca.shape
```

```python
component_df=pd.DataFrame(data=df_pca,columns=["first component","second component"])
```

```python
component_df
```

```python
pca.explained_variance_
```

```python
pca.explained_variance_ratio_
```

```python
np.cumsum(pca.explained_variance_ratio_)
```

PCA with four components

```python
pca= PCA(n_components=4)
```

```python
df_pca=pca.fit_transform(df_new)
```

```python
component_df=pd.DataFrame(data=df_pca,columns=["first component","second component","third component","fourth component"])
```

```python
component_df
```

```python
pca.explained_variance_
```

```python
pca.explained_variance_ratio_
```

```python
np.cumsum(pca.explained_variance_ratio_)
```
