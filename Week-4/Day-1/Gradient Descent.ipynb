Splitting into training and testing data

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df[['age','affordibility']],df.bought_insurance,test_size=0.2,random_state=25)
```

Data scaling

```python
X_train_scaled= X_train.copy()
X_train_scaled['age']=X_train_scaled['age']/100

X_test_scaled=X_test.copy()
X_test_scaled['age']=X_test_scaled['age']/100
```

Making the Neural Network using Tensorflow

```python
model= keras.Sequential([
    keras.layers.Dense(1,input_shape=(2,), activation='sigmoid', kernel_initializer='ones',bias_initializer='zeros')
])
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.fit(X_train_scaled,y_train,epochs=5000)
```

```python
model.evaluate(X_test_scaled,y_test)
```

```python
model.predict(X_test_scaled) #Anything is more than 0.5,the person will buy the insurance
```

Calculating the Output using the sigmoid function

```python
def sigmoid(x):
    import math
    return 1/ (1 + np.exp(-x))
```

```python
def prediction_function(age,affordibility):
    weighted_sum=coef[0]*age+coef[1]*affordibility+intercept
    return sigmoid(weighted_sum)
```

Creating a Gradient Descent Function using Python from scratch

```python
def log_loss(y_true,y_prediction):
    epsilon= 1e-15
    y_prediction_new = [max(i,epsilon) for i in y_prediction]
    y_prediction_new = [min(i,1-epsilon) for i in y_prediction_new]
    y_prediction_new = np.array(y_prediction_new)
    return -np.mean(y_true*np.log(y_prediction_new)+(1-y_true)*np.log(1-y_prediction_new))
```

```python
def sigmoid_numpy(x):
    return 1/(1+np.exp(-x))
sigmoid_numpy(np.array([12,0,1]))
```

```python
def gradient_descent(age,affordibility,y_true,epochs):
    #Will return w1,w2,bias
    w1 = w2 = 1
    bias = 0
    learning_rate= 0.5
    n=len(age)
    
    
    for i in range(epochs):
        weighted_sum=w1 * age + w2 + affordibility + bias
        y_predicted=sigmoid_numpy(weighted_sum)
        loss = log_loss(y_true,y_predicted)

        w1d=(1/n)*np.dot(np.transpose(age),(y_predicted-y_true))
        w2d=(1/n)*np.dot(np.transpose(affordibility),(y_predicted-y_true))
        bias_d=np.mean(y_predicted-y_true)

        w1=w1-learning_rate*w1d
        w2=w2-learning_rate*w2d
        bias = bias - learning_rate*bias_d
        
        print(f"Epoch:{i},w1:{w1},w1:{w2},bias:{bias},loss:{loss}")    

    return w1,w2,bias
```
