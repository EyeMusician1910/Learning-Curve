Making the training and testing data

```python
X= df.drop(60,axis='columns')
y=df[60]
y.head()
```

```python
#Converting the target calues into binary values
y=pd.get_dummies(y,drop_first=True)
y.sample(5)
```

```python
y.value_counts()
```

splitting into training and Testing Data

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=1)
```

```python
import tensorflow as tf
from tensorflow import keras
```

Building the Neural Network

```python
model = keras.Sequential([
    keras.layers.Dense(60,activation='relu'),#3 hidden layers with 60,30 and 15 neurons respectively
    keras.layers.Dense(30,activation='relu'),
    keras.layers.Dense(15,activation='relu'),
    keras.layers.Dense(1,activation='sigmoid')#Output layer with sigmoid function as we need binary classification
])

model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])#compliling the model

model.fit(X_train,y_train,epochs=100,batch_size=8)#training the model with 100 epochs and the batch size of 8,we're using the mini batch gradient descent,we feed 8 samples in each itereations and then do a backpropogation
#We purposefully overfitted the model to use the dropout regularization technique to reduce the overfitting
```

```python
model.evaluate(X_test,y_test)
```

```python
#making some predictions
y_pred=model.predict(X_test).reshape(-1)
print(y_pred[:10])

#round the values to nearest to 0 and 1 
y_pred=np.round(y_pred)
print(y_pred[:10])
```

```python
y_test[:10]
```

Creating a classification report for the more detailed approach

```python
from sklearn.metrics import confusion_matrix,classification_report
print(classification_report(y_test,y_pred))
```

making a model with a dropout layer

```python
modeld = keras.Sequential([
    keras.layers.Dense(60,activation='relu'),#3 hidden layers with 60,30 and 15 neurons respectively
    keras.layers.Dropout(0.5),#dropout with  50% dropout rate,it'll drop 50% of the neurons in the layer randomly
    keras.layers.Dense(30,activation='relu'),
    keras.layers.Dropout(0.5),#dropout with  50% dropout rate,it'll drop 50% of the neurons in the layer randomly
    keras.layers.Dense(15,activation='relu'),
    keras.layers.Dropout(0.5),#dropout with  50% dropout rate,it'll drop 50% of the neurons in the layer randomly
    keras.layers.Dense(1,activation='sigmoid')#Output layer with sigmoid function as we need binary classification
])

modeld.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])#compliling the model

modeld.fit(X_train,y_train,epochs=100,batch_size=8)#training the model with 100 epochs and the batch size of 8,we're using the mini batch gradient descent,we feed 8 samples in each itereations and then do a backpropogation
#We purposefully overfitted the model to use the dropout regularization technique to reduce the overfitting
```

```python
modeld.evaluate(X_test,y_test)
```

```python
y_pred=modeld.predict(X_test).reshape(-1)
y_pred=np.round(y_pred)
print(classification_report(y_test,y_pred))
```
